```python
# Aula: Instrumentação automática
# Experimento Terra-Água
# Data: 26/11/2026
# Aluno: José Figueiredo
#----------------------------------------------------------------------

#1. Abrir arquivos de dados do experimento
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

url='https://raw.githubusercontent.com/jlafigueiredo2/meteorologia-pub/refs/heads/main/Exp_exp01.dat'

#2. leitura do arquivo
df =pd.read_csv(url,header=1,skiprows=[2,3])
df
```

```python
#3. Definição das variáveis de máximo e mínimo e suas posições (índices)
ti_a = np.min(df['t_agua_Avg'])
ti_t = np.min(df['t_terra_Avg'])
tf_a = np.max(df['t_agua_Avg'])
tf_t = np.max(df['t_terra_Avg'])

ti_ax = np.argmin(df['t_agua_Avg'])
ti_tx = np.argmin(df['t_terra_Avg'])
tf_ax = np.argmax(df['t_agua_Avg'])
tf_tx = np.argmax(df['t_terra_Avg'])

#4. Configurações do gráfico
plt.rcParams['figure.dpi'] = 300
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.size'] = 11

fig, ax = plt.subplots(figsize=(8.5, 6))

#5. Linhas temperatura da água e terra (espessura e cores)
ax.plot(df['t_agua_Avg'], label='Água ($c_p = 1,00$ cal/g°C)', color='#1f77b4', linewidth=2.2)
ax.plot(df['t_terra_Avg'], label='Terra ($c_p = 0,47$ cal/g°C)', color='#d62728', linewidth=2.2)

#6. Pontos Mínimo, Máximo, e posição na tabela - Água
ax.scatter(ti_ax, ti_a, s=70, color='#1f77b4', edgecolors='black', zorder=5)
ax.text(ti_ax + 10, ti_a - 0.2, f'{ti_a:.2f}°C', color='#1f77b4', fontweight='bold', ha='left', va='center')

ax.scatter(tf_ax, tf_a, s=70, color='#1f77b4', edgecolors='black', zorder=5)
ax.text(tf_ax, tf_a + 0.6, f'{tf_a:.2f}°C', color='#1f77b4', fontweight='bold', ha='center', va='bottom')

#7. Pontos Mínimo, Máximo, e posição na tabela - Terra
ax.scatter(ti_tx, ti_t, s=70, color='#d62728', edgecolors='black', zorder=5)
ax.text(ti_tx - 10, ti_t + 0.5, f'{ti_t:.2f}°C', color='#d62728', fontweight='bold', ha='right', va='center')

ax.scatter(tf_tx, tf_t, s=70, color='#d62728', edgecolors='black', zorder=5)
ax.text(tf_tx + 12, tf_t, f'{tf_t:.2f}°C', color='#d62728', fontweight='bold', ha='left', va='center')

#8. Títulos do Gráfico
titulo_negrito = "EXPERIMENTO TERRA-ÁGUA\nAnálise da curva térmica"
ax.set_title(titulo_negrito, fontsize=12, fontweight='bold', pad=25, loc='center')
ax.text(0.5, 1.03, "Florianópolis 26/11/2026", fontsize=10, fontweight='normal', ha='center', va='bottom', transform=ax.transAxes)

#9. Rótulo dos X e Y  e limite
ax.set_xlabel('Tempo [30 segundos]', fontsize=11, labelpad=10)
ax.set_ylabel('Temperatura °C', fontsize=11, labelpad=10)
ax.set_ylim(22.0, 38.0)

#10. Grade e Legenda
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(
    title='Calor Específico',
    title_fontsize=11,
    loc='upper right',
    frameon=True,
    facecolor='#fcfcfc',
    edgecolor='#cccccc'
)
plt.tight_layout()
plt.show()
```
